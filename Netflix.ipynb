**we need to fix the issue. rating colum has minutes which does not belong there**

**Step 5: Fix the issue **

```python
# find which rows have the problem
mask = df["rating"].str.contains("min", na=False)
df[mask][["title", "rating", "duration"]]
```

**Step 6: Move the value to correct column**

```python
# move the value from rating column to duration column
df.loc[mask, "duration"] = df.loc[mask, "rating"]

# clear the wrong value from rating colum
df.loc[mask, "rating"] = None

# confirm the fix
df[mask][["title", "rating", "duration"]]
```

**Step 7: Double check if its fixed**

```python
df['rating'].unique()
```

**Step 8: Investigat missing values**

```python
df.isnull().sum()
```

**Step 9: Decide How to Handle Each Missing Column**

```python
# fill director, cast, country with "Unknown"
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
```

**Step 10: Delete rws with missing date_added or rating**

```python
df = df.dropna(subset=["date_added", "rating"])
```

**STEP 11: Confirm everything is fixed**

```python
df.isnull().sum()
```

**Step 14: Convert to realdate**

```python
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')

# confirm if it worked
df['date_added'].dtype
df['date_added'].head()
```

**Step 15: Extract Year and Month**

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month

df[['date_added', 'year_added', 'month_added']].head()
```

**Step 16:  Investigate and Split the duration Column**

```python
# check whats inside duration
df['duration'].unique()
```

```python
# create duration_mins for movies
df.loc[df["type"] == "Movie", "duration_mins"] = df[df["type"] == "Movie"]["duration"].str.replace(" min", "", regex=False).astype(int)
df[df["type"] == "Movie"]["duration_mins"] = df[df["type"] == "Movie"]["duration_mins"].astype(int)
# display the head of the relevant columns
df[df["type"] == "Movie"][["title","duration", "duration_mins"]].head()
```

```python
# create duration_seasons for TV Shows
df['duration_seasons'] = df[df['type'] == 'TV Show']['duration'].str.replace(' Seasons', '', regex=False).str.replace(' Season', '', regex=False).astype(int)
# display head of the relevant column
df[df['type'] == 'TV Show'][['title', 'duration', 'duration_seasons']].head()
```
